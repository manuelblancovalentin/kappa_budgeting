## Precision Dictionary

```python
from enabol import dtypes, PrecisionDict

precisions = PrecisionDict({
    "input": {
        "value": "ap_fixed<12,4,AP_RND,AP_SAT>",
    },
    "dense0": {
        "weight": dtypes.ap_fixed(12, 4, "AP_RND", "AP_SAT"),
        "gradient": dtypes.ap_fixed(16, 6, "AP_RND", "AP_SAT"),
        "update": dtypes.ap_fixed(16, 4, "AP_RND", "AP_SAT"),
    },
    "loss": {
        "value": "ap_fixed<24,12,AP_RND,AP_SAT>",
    },
})
print(precisions)
```

```python
t = dtypes.ap_fixed(8, 3, "AP_RND", "AP_SAT")
print(t)
real_values = t([0.1, 0.2, 12.0])
raw_ints = t([0.1, 0.2, 12.0], return_int=True)
raw_ints
```

```python
import numpy as np
W = np.array([[0.1, 0.2], [-0.1, 0.4]], dtype=np.float32)
weight_dtype = dtypes.ap_fixed(12, 4, "AP_RND", "AP_SAT")
W_q = weight_dtype(W)
W_q
```

```python
W_b = weight_dtype.bin(W)
print(W_b)
```
